```python
import pandas as pd
```

```python
# Cleaned tables
customers = pd.read_csv("customers_cleaned.csv")
billing = pd.read_csv("billing_cleaned.csv")
subscriptions = pd.read_csv("subscriptions_cleaned.csv")
payments = pd.read_csv("payments_cleaned.csv")

# Original tables
plans = pd.read_csv("plans.csv")
plan_history = pd.read_csv("plan_history.csv")
contracts = pd.read_csv("contracts.csv")
devices = pd.read_csv("devices.csv")
recharges = pd.read_csv("recharges.csv")
usage_voice = pd.read_csv("usage_voice.csv")
usage_sms = pd.read_csv("usage_sms.csv")
usage_data = pd.read_csv("usage_data.csv")
network_quality = pd.read_csv("network_quality.csv")
support_tickets = pd.read_csv("support_tickets.csv")
complaints = pd.read_csv("complaints.csv")
customer_feedback = pd.read_csv("customer_feedback.csv")
retention_campaigns = pd.read_csv("retention_campaigns.csv")
marketing_campaigns = pd.read_csv("marketing_campaigns.csv")
employees = pd.read_csv("employees.csv")
stores = pd.read_csv("stores.csv")
regions = pd.read_csv("regions.csv")
cities = pd.read_csv("cities.csv")
states = pd.read_csv("states.csv")
data_quality_issue_log = pd.read_csv("data_quality_issue_log.csv")
```

```python
# Customers
customers["date_of_birth"] = pd.to_datetime(customers["date_of_birth"])
customers["acquisition_date"] = pd.to_datetime(customers["acquisition_date"])
customers["churn_date"] = pd.to_datetime(customers["churn_date"])

# Billing
billing["billing_date"] = pd.to_datetime(billing["billing_date"])
billing["due_date"] = pd.to_datetime(billing["due_date"])

# Complaints
complaints["complaint_date"] = pd.to_datetime(complaints["complaint_date"], format="%d-%m-%Y")
complaints["resolution_date"] = pd.to_datetime(complaints["resolution_date"], format="%d-%m-%Y")

# Contracts
contracts["start_date"] = pd.to_datetime(contracts["start_date"], format="%d-%m-%Y")
contracts["end_date"] = pd.to_datetime(contracts["end_date"], format="%d-%m-%Y")

# Customer Feedback
customer_feedback["feedback_date"] = pd.to_datetime(customer_feedback["feedback_date"], format="%d-%m-%Y")

# Payments
payments["payment_date"] = pd.to_datetime(payments["payment_date"])

# Devices
devices["purchase_date"] = pd.to_datetime(devices["purchase_date"], format="%d-%m-%Y")

# Employees
employees["hire_date"] = pd.to_datetime(employees["hire_date"], format="%d-%m-%Y")

# Marketing Campaigns
marketing_campaigns["start_date"] = pd.to_datetime(marketing_campaigns["start_date"], format="%d-%m-%Y")
marketing_campaigns["end_date"] = pd.to_datetime(marketing_campaigns["end_date"], format="%d-%m-%Y")

# Plan History
plan_history["change_date"] = pd.to_datetime(plan_history["change_date"], format="%d-%m-%Y")

# Recharges
recharges["recharge_date"] = pd.to_datetime(recharges["recharge_date"], format="%d-%m-%Y")

# Retention Campaigns
retention_campaigns["contact_date"] = pd.to_datetime(retention_campaigns["contact_date"], format="%d-%m-%Y")

# Subscriptions
subscriptions["start_date"] = pd.to_datetime(subscriptions["start_date"])
subscriptions["end_date"] = pd.to_datetime(subscriptions["end_date"])

# Stores
stores["opened_date"] = pd.to_datetime(stores["opened_date"], format="%d-%m-%Y")

# Support Tickets
support_tickets["created_date"] = pd.to_datetime(support_tickets["created_date"], format="%d-%m-%Y")
support_tickets["resolution_date"] = pd.to_datetime(support_tickets["resolution_date"], format="%d-%m-%Y")
```

```python
# KPI 1: Customer Churn Rate
```

```python
churned_customers = customers[customers["customer_status"] == "Churned"]
Total_customers = customers["customer_status"].count()

churn_rate = round(len(churned_customers) / Total_customers*100,2)
print("Customer Churn Rate:", churn_rate,"%")
```

```python
# KPI 2: Customer Retention Rate
```

```python
Retained_customers = customers[customers["customer_status"] == "Active"]
Total_customers = customers["customer_status"].count()

Retention_rate = round(len(Retained_customers) / Total_customers*100,2)
print("Customer Retention Rate:", Retention_rate,"%")
```

```python
# KPI 3: ARPU
```

```python
Total_Recurring_Revenue = subscriptions[subscriptions["subscription_status"] == "Active"]["monthly_charge_inr"].sum()
Avg_Active_Subscribers =  subscriptions[subscriptions["subscription_status"] == "Active"]

ARPU = round(Total_Recurring_Revenue / len(Avg_Active_Subscribers),2)
print("ARPU:", ARPU)
```

```python
# KPI 4: Customer Lifetime Value
```

```python
Gross_Margin = 0.45
CLV = round(ARPU * Gross_Margin * (1 / (churn_rate / 100)),2)

print("Customer Lifetime Value (CLV): ₹", CLV)
```

**Assumption**: Gross Margin data(Cost Column not available) was not available in the dataset. Therefore, a **45% Gross Margin** was assumed for the CLV calculation as per Telecom Industry Benchmark.

```python
# KPI 5: Revenue Lost to Churn
```

```python
churned_customers = len(customers[customers["customer_status"] == "Churned"])

RLC = churned_customers * ARPU
print("Revenue Lost to Churn: ₹", RLC)
```

```python
# KPI 6: Monthly Recurring Revenue
```

```python
Monthly_Recurring_Revenue = subscriptions[subscriptions["subscription_status"] == "Active"]["monthly_charge_inr"].sum()
print("Monthly Recurring Revenue: ₹", Monthly_Recurring_Revenue)
```

```python
#KPI 7: First Contact Resolution
```

```python
Resolved_First_Contact = support_tickets[support_tickets["first_contact_resolution"] == "Yes"]
Total_Tickets = len(support_tickets)

FCR = round((len(Resolved_First_Contact) / Total_Tickets) * 100, 2)
print("First Contact Resolution (FCR):", FCR, "%")
```

```python
#KPI 8: Avg Resolution Time
```

```python
resolved_tickets = support_tickets[support_tickets["resolution_date"].notna()].copy()

resolved_tickets["calculated_hours"] = (resolved_tickets["resolution_date"] - resolved_tickets["created_date"]).dt.total_seconds() / 3600
Avg_Resolution_Time = round(resolved_tickets["calculated_hours"].mean(), 2)

print("Average Resolution Time:", Avg_Resolution_Time, "hours")
```

```python
#KPI 9: Contract Renewal Rate
```

```python
Contracts_Renewed = len(contracts[contracts["renewal_status"].isin(["Renewed", "Auto-Renewed"])])
Due_for_Renewal = len(contracts[contracts["renewal_status"].isin(["Renewed", "Auto-Renewed", "Pending", "Not Renewed"])])

CRR = round(Contracts_Renewed / Due_for_Renewal*100,2)
print("Contract Renewal Rate:", CRR, "%")
```

```python
#KPI 10: Avg Customer Tenure	
```

```python
ACT = round(customers[customers["customer_status"] == "Active"]["tenure_months"].mean(),2)
print("Avg Customer Tenure:", ACT, "months")
```
