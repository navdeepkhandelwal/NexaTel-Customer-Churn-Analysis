```python
Q1 = usage_data["data_gb_used"].quantile(0.25)
Q3 = usage_data["data_gb_used"].quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - (1.5 * IQR)
upper_limit = Q3 + (1.5 * IQR)

outliers = usage_data[(usage_data["data_gb_used"] < lower_limit) |
                      (usage_data["data_gb_used"] > upper_limit)]

print("Number of Outliers :", len(outliers))
outliers
```

```python
total_amount_inr_negative_values = billing[billing["total_amount_inr"] < 0]
print("Negative values in total_amount_inr :", len(total_amount_inr_negative_values))
total_amount_inr_negative_values
```

## Observation

- A total of **4,480 outliers** were identified in the **data_gb_used** column using the IQR method. These values have been flagged for further investigation to determine whether they represent genuine high data usage or potential data quality issues.

- A total of **229 records** contain negative values in the **total_amount_inr** column. Since billing amounts are expected to be non-negative, these records have been flagged as potential data quality issues and will be reviewed during the data cleaning process.
